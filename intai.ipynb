**선형 코드**

```python
import numpy as np
import matplotlib.pyplot as plt

# 키(cm)와 몸무게(kg) 데이터
X = np.array([150, 155, 160, 165, 170, 175, 180], dtype=float)
y = np.array([45, 50, 54, 58, 63, 68, 73], dtype=float)

W = 0           # 기울기
b = 0           # 절편
lrate = 0.00001 # 학습률
epochs = 10000  # 반복 횟수
n = float(len(X))

# 경사하강법
for i in range(epochs):

    # 예측값
    y_pred = W * X + b

    # W와 b의 변화량 계산
    dW = (2 / n) * sum(X * (y_pred - y))
    db = (2 / n) * sum(y_pred - y)

    # W와 b 수정
    W = W - lrate * dW
    b = b - lrate * db

# 최종 기울기와 절편
print("Final W =", W)
print("Final b =", b)

# 키가 165cm일 때 몸무게 예측
height = 165
weight_pred = W * height + b

print("키가", height, "cm일 때 예상 몸무게 =", weight_pred, "kg")

# 학습된 직선의 예측값
y_pred = W * X + b

# 그래프
plt.scatter(X, y)
plt.plot(X, y_pred, color='red')
plt.scatter(height, weight_pred, color='blue')

plt.xlabel("Height (cm)")
plt.ylabel("Weight (kg)")
plt.title("Height and Weight Linear Regression")

plt.show()
```
